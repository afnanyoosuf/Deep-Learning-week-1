```python
import numpy as np 
import torch 
from torchvision import datasets,transforms
 
```

```python
transform = transforms.ToTensor()

train_dataset = datasets.MNIST(
    root="data/raw",
    train=True,
    download=True,
    transform=transform
)
```

```python
Image,label = train_dataset[0]

print("image shape:",Image.shape)
print("Label:",label)
```

```python
x = Image.numpy()
print(x.shape)
```

```python
x = x.reshape(1,784)
print(x.shape)
```

```python
np.random.seed(42)

W1 = np.random.randn(784, 128) * 0.01
b1 = np.zeros((1, 128))

W2 = np.random.randn(128, 10) * 0.01
b2 = np.zeros((1, 10))
```

```python
print("W1:", W1.shape)
print("b1:", b1.shape)

print("W2:", W2.shape)
print("b2:", b2.shape)
```

```python
def forward_propagation(x, W1, b1, W2, b2):

    # First layer
    Z1 = x @ W1 + b1

    # ReLU
    A1 = np.maximum(0, Z1)

    # Output layer
    Z2 = A1 @ W2 + b2

    return Z2
```

```python
output = forward_propagation(
    x,
    W1,
    b1,
    W2,
    b2
)

print("Output shape:", output.shape)
print("Output:", output)
```
