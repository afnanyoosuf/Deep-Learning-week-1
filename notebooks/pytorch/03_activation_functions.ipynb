```python
import numpy as np 
import matplotlib.pyplot as plt
```

```python
#Define the activation functions

def relu(x):
    return np.maximum(0,x)

def sigmoid(x):
    return 1 /(1 + np.exp(-x))

def softmax(x):
    exp_x = np.exp(x-np.max(x))
    return exp_x /np.sum(exp_x)

```

```python
x = np.linspace(-10,10,500)
```

```python
y_relu = relu(x)

plt.figure(figsize=(8,5))

plt.plot(x,y_relu)

plt.axhline(0, linewidth = 0.8)
plt.axvline(0, linewidth = 0.8)

plt.title("ReLU Activation Function")
plt.xlabel("Input")
plt.ylabel("ReLu(x)")

plt.grid(True)
plt.show()
```

```python
y_sigmoid = sigmoid(x)

plt.figure(figsize=(8,5))

plt.plot(x,y_sigmoid)

plt.axhline(0, linewidth = 0.8)
plt.axhline(0.5, linestyle ="--", linewidth = 0.8 )
plt.axhline(1,linewidth = 0.8)
plt.axvline(0, linewidth =0.8)

plt.title("Sigmoid Activation Function")
plt.xlabel("Input")
plt.ylabel("sigmoid(x)")

plt.grid(True)
plt.show()
```

```python
logits = np.array([2.0,1.0,0.5,3.0,1.5])

probabilites = softmax(logits)

print("Logitis:",logits)
print("Probabilites:",probabilites)
print("Sum:",probabilites.sum())

```

```python
classes = ["0","1","2","4","5"]

plt.figure(figsize=(8,5))

plt.bar(classes,probabilites)

plt.title("Softmax Output")
plt.xlabel("Class")
plt.ylabel("Probility")

plt.ylim(0,1)
plt.grid(axis="y")

```

```python
prediction = np.argmax(probabilites)

print("Predicted Class",prediction)
```

```python
# Compare ReLU and Sigmoid

plt.figure(figsize=(9,5))

plt.plot(x,relu(x),label = "ReLu")
plt.plot(x,sigmoid(x),label = "Sigmoid")

plt.axhline(0,linewidth = 0.8)
plt.axvline(0,linewidth = 0.8)

plt.title("Activation Function")
plt.xlabel("Input")
plt.ylabel("Output")

plt.legend()
plt.grid(True)
```
